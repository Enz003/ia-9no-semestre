# Ejercicio 2

**Definimos espacio**

```python
def definirEspacio(filas,columnas):
    espacio = [[True for _ in range(columnas)] for _ in range(filas)]
    return espacio
```

**Definimos Heuristica**

```python
def heuristica(nodoInicio: tuple, nodoFin: tuple):
    absInicio = nodoInicio[0]
    absFin = nodoFin[0]
    ordInicio = nodoInicio[1]
    ordFin = nodoFin[1]
    h = ((ordFin-ordInicio)**2 + (absFin-absInicio)**2)**0.5
    return float(h)
```

**Definicion de obstaculos**

```python
def definirObstaculos(obstaculos: list, espacio):
    for i in obstaculos:
        espacio[i[1]-1][i[0]-1] = False
    
```
